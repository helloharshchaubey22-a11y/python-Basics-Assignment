```python
#6.  Write a function to demonstrate the behavior of mutable and immutable arguments.
```

```python
# Function to demonstrate mutable and immutable behavior
def modify_values(num, lst):
    # Trying to change the integer (immutable)
    num = num + 10

    # Trying to change the list (mutable)
    lst.append(100)

    print("Inside function - num:", num)
    print("Inside function - lst:", lst)


# Main program
x = 50                  # Immutable (integer)
my_list = [1, 2, 3]     # Mutable (list)

print("Before function call:")
print("x:", x)
print("my_list:", my_list)

modify_values(x, my_list)

print("\nAfter function call:")
print("x:", x)                # Will NOT change
print("my_list:", my_list)    # Will change
```

```python
#7.  Write a program that performs basic arithmetic operations on two user-input numbers.
```

```python
# Program to perform basic arithmetic operations on two numbers

# Taking user input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Performing operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2

# To avoid division by zero
if num2 != 0:
    division = num1 / num2
else:
    division = "Undefined (cannot divide by zero)"

# Displaying results
print("\n--- Results ---")
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

```python
#9. Write a Python program to convert user input from string to integer, float, and boolean types.
# Program to convert user input into different data types

# Taking input as a string
user_input = input("Enter any value: ")

# Convert to integer (only works if input is a valid integer)
try:
    int_value = int(user_input)
except ValueError:
    int_value = "Cannot convert to integer"

# Convert to float (works if input is a number)
try:
    float_value = float(user_input)
except ValueError:
    float_value = "Cannot convert to float"

# Convert to boolean
# In Python: empty string = False, any non-empty string = True
bool_value = bool(user_input)

# Displaying results
print("\n--- Converted Values ---")
print("Original (string):", user_input)
print("Integer value:", int_value)
print("Float value:", float_value)
print("Boolean value:", bool_value)
```

```python
#10 Write code to demonstrate type casting with list elements.
# List containing string numbers
string_list = ["10", "20", "30", "40"]

print("Original list (strings):", string_list)

# Type casting each element to integer
int_list = [int(x) for x in string_list]
print("List after casting to integers:", int_list)

# Type casting each element to float
float_list = [float(x) for x in string_list]
print("List after casting to floats:", float_list)

# Type casting each element to boolean
# Note: Any non-empty string becomes True
bool_list = [bool(x) for x in string_list]
print("List after casting to booleans:", bool_list)
```

```python
#13.Write a Python program to find the sum of all even numbers between 1 and 50.
# Program to find the sum of all even numbers between 1 and 50

total = 0

for num in range(1, 51):
    if num % 2 == 0:   # Check if the number is even
        total += num

print("The sum of even numbers from 1 to 50 is:", total)
```

```python
#14.Write a program to reverse a string using a while loop.
# Program to reverse a string using a while loop

# Taking user input
text = input("Enter a string: ")

# Variables for reversing
reversed_text = ""
index = len(text) - 1

# Using while loop to reverse
while index >= 0:
    reversed_text += text[index]
    index -= 1

# Output
print("Reversed string:", reversed_text)
```

```python
#15 Write a Python program to calculate the factorial of a number provided by the user using a while loop.
# Program to calculate factorial of a number using a while loop

num = int(input("Enter a number: "))

factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print("Factorial of", num, "is:", factorial)
```
